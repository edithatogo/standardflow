# PRISMA Diagram Generator

This notebook demonstrates how to generate a PRISMA 2020 flow diagram using the `PrismaDiagram` class.

```python
import sys
import json
sys.path.append('../src')
from prisma import PrismaDiagram
from IPython.display import Image, SVG
```

## Load Data from a JSON file

```python
with open('../data/prisma_data.json', 'r') as f:
    data = json.load(f)
```

## Create a PrismaDiagram object

```python
prisma = PrismaDiagram(data)
```

## Generate a PNG Diagram

```python
prisma.generate('../prisma_generated.png')
Image('../prisma_generated.png')
```

## Generate an SVG Diagram

```python
prisma.generate('../prisma_generated.svg')
SVG('../prisma_generated.svg')
```
