# CONSORT 2010 Cluster-Randomised Trial Diagram Example

This notebook demonstrates how to generate a CONSORT 2010 flow diagram for a cluster-randomised trial using the `ConsortDiagram` class.

```python
import sys
import json
sys.path.append('../../src')
from consort import ConsortDiagram
from IPython.display import Image
```

## Load Data from a JSON file

```python
with open('../../data/consort_cluster_data.json', 'r') as f:
    data = json.load(f)
```

## Create a ConsortDiagram object

```python
consort = ConsortDiagram(data)
```

## Generate and Display the Diagram

```python
output_filename = '../images/consort_cluster_example_from_notebook.png'
consort.generate(output_filename)
Image(output_filename)
```
