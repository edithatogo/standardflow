# PRISMA 2020 Updated Diagram Example

This notebook demonstrates how to generate a PRISMA 2020 flow diagram for an updated review using the `PrismaDiagram` class.

```python
import sys
import json
sys.path.append('../../src')
from prisma import PrismaDiagram
from IPython.display import Image
```

## Load Data from a JSON file

```python
with open('../../data/prisma_2020_updated_data.json', 'r') as f:
    data = json.load(f)
```

## Create a PrismaDiagram object

```python
prisma = PrismaDiagram(data)
```

## Generate and Display the Diagram

```python
output_filename = '../images/prisma_2020_updated_example_from_notebook.png'
prisma.generate(output_filename)
Image(output_filename)
```
