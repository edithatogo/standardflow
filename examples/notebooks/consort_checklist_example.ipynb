# CONSORT Checklist Example

This notebook demonstrates how to generate a CONSORT 2010 checklist using the `Checklist` class.

```python
import sys
import json
sys.path.append('../../src')
from checklist import Checklist
from IPython.display import Markdown
```

## Load Data from a JSON file

```python
with open('../../data/consort_checklist.json', 'r') as f:
    data = json.load(f)
```

## Create a Checklist object

```python
checklist = Checklist(data)
```

## Generate and Display the Checklist

```python
markdown = checklist.to_markdown()
display(Markdown(markdown))
```
