# STROBE Cohort Study Diagram Example

This notebook demonstrates how to generate a STROBE flow diagram for a cohort study using the `StrobeDiagram` class.

```python
import sys
import json
sys.path.append('../../src')
from strobe import StrobeDiagram
from IPython.display import Image
```

## Load Data from a JSON file

```python
with open('../../data/strobe_cohort_data.json', 'r') as f:
    data = json.load(f)
```

## Create a StrobeDiagram object

```python
strobe = StrobeDiagram(data)
```

## Generate and Display the Diagram

```python
output_filename = '../images/strobe_cohort_example_from_notebook.png'
strobe.generate(output_filename)
Image(output_filename)
```
